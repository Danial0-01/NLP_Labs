## 1. Load a Transformer

We use **DistilBERT**, a  Transformer fine-tuned for positive/negative sentiment classification.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.eval()

# Look inside the Transformer architecture.
print("Transformer layers:", model.config.n_layers)
print("Attention heads per layer:", model.config.n_heads)
print("Hidden size:", model.config.dim)
```


<img src="https://drive.google.com/thumbnail?id=1v2dRK1KR_ydZOknjj8Xv8_wjOdXleDYs&sz=w1000" width="500">

## 2. Tokenizing the text

```python
sentence = "The movie was not good."

# Tokenization: converts text into token IDs
inputs = tokenizer(sentence, return_tensors="pt")

# [CLS] = start token, [SEP] = end token
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

print("Tokens:", tokens)
```

## 3. Embeddings and self-attention

Each token ID becomes a **vector (embedding)**. Inside each Transformer layer, **self-attention** lets a token use information from other tokens in the same sentence.

```python
import torch

# Enable attention weights
model.set_attn_implementation("eager")

with torch.no_grad():
    embeddings = model.distilbert.embeddings(inputs["input_ids"])
    outputs = model(**inputs, output_attentions=True)

# Attention from the first layer and first head
attention = outputs.attentions[0][0, 0]

print("Embeddings shape:", embeddings.shape)
print("Attention shape:", attention.shape)
print("Tokens:", tokens)
print("Attention weights:\n", attention)
```

## 4. Classification: From Transformer to prediction

The Transformer produces contextual representations. The **classification head** turns them into two scores (*logits*). **Softmax** converts these into probabilities that sum to 1.

```python
# Get prediction scores from the Transformer
with torch.no_grad():
    logits = model(**inputs).logits

# Convert scores into probabilities
probs = torch.softmax(logits, dim=1)[0]

# Display the prediction
prediction = probs.argmax().item()

print("Prediction:", model.config.id2label[prediction])
print("Confidence:", round(probs[prediction].item(), 3))
```

#### Task 1: Number of Transformer layers and attention heads per layer

```python
from transformers import AutoTokenizer, AutoModel
import torch

bert_name = "bert-base-uncased"
bert_tokenizer = AutoTokenizer.from_pretrained(bert_name)
bert_model = AutoModel.from_pretrained(bert_name)
bert_model.set_attn_implementation("eager")   # needed to get attention weights back
bert_model.eval()

num_layers = bert_model.config.num_hidden_layers
num_heads  = bert_model.config.num_attention_heads

print("Number of Transformer layers:", num_layers)
print("Attention heads per layer   :", num_heads)
```

#### Task 2: Total number of attention heads

```python
total_heads = num_layers * num_heads
print(f"Total attention heads = {num_layers} layers x {num_heads} heads = {total_heads}")
```

#### Task 3: Attention weights from the 3rd layer, 2nd attention head

Python counts from 0, so the **3rd layer is index 2** and the **2nd head is index 1**. `outputs.attentions` is a tuple with one entry per layer, and each entry has shape `(batch, heads, seq_len, seq_len)`.

```python
import pandas as pd
import matplotlib.pyplot as plt

sentence = "Transformers use self-attention to understand language."
bert_inputs = bert_tokenizer(sentence, return_tensors="pt")
bert_tokens = bert_tokenizer.convert_ids_to_tokens(bert_inputs["input_ids"][0])

with torch.no_grad():
    bert_outputs = bert_model(**bert_inputs, output_attentions=True)

# layer index 2 = 3rd layer, [batch 0, head index 1 = 2nd head]
attention_l3_h2 = bert_outputs.attentions[2][0, 1]

print("Tokens:", bert_tokens)
print("Attention matrix shape:", tuple(attention_l3_h2.shape))
print("Each row sums to 1:", torch.allclose(attention_l3_h2.sum(dim=-1), torch.ones(len(bert_tokens))))

attention_df = pd.DataFrame(attention_l3_h2.numpy(), index=bert_tokens, columns=bert_tokens).round(3)
display(attention_df)

# Heatmap
plt.figure(figsize=(7, 6))
plt.imshow(attention_l3_h2.numpy(), cmap="Blues")
plt.xticks(range(len(bert_tokens)), bert_tokens, rotation=90)
plt.yticks(range(len(bert_tokens)), bert_tokens)
plt.xlabel("Token being attended to (column)")
plt.ylabel("Token doing the attending (row)")
plt.title("BERT attention - layer 3, head 2")
plt.colorbar()
plt.tight_layout()
plt.show()
```

#### Task 4: What do the rows and columns of the attention matrix represent?

The attention matrix is square, with one row and one column for every token in the input, including the special `[CLS]` and `[SEP]` tokens (shape `seq_len x seq_len`).

- **Rows = the token that is "looking".** Row *i* describes the token at position *i* as it builds its new, context-aware representation (the *query* token).
- **Columns = the tokens being "looked at".** Column *j* is the token at position *j* that row *i* may take information from (the *key* token).
- **Each value `attention[i][j]`** is how much token *i* attends to token *j*. It is a probability: the scores go through a softmax, so every value is between 0 and 1 and **each row adds up to 1**. A large value means token *i* draws a lot of information from token *j* when it updates its representation.

Two further points: the matrix is generally **not symmetric**, because token *i* attending to token *j* does not imply the reverse; and every attention head in every layer has its own matrix, so this one only shows the pattern of one head (head 2 of layer 3). BERT is bidirectional, so each token can attend to tokens both before and after it, and a token may also attend to itself (the diagonal).
